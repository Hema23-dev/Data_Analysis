## 1.2 Pull up the average for insurance charges based on region

Using groupby function , we are grouping it with region and we need charges average so taken charges.mean() adn then we need the value to 

be rounnded of to 2 digit so round(2) adn sorting it in ascending order so  using sort_values() here we need most charges paid , keeping ascending=False


```python
insurance_data["region"].unique()  # to find the unique element in that particular column
```

```python
insurance_data.groupby("region")["charges"].mean().round(2).sort_values(ascending=False)
```

## 1.3 Pull up the average insurance charges based on region and sex

* Discrete(Categorial) - it is countable and not measurable and not segmented

* continous - it is measurable and have unit and can be segmented

*** important ****

So here region adn sex will be taken in groupby because it is discrete , outside of groupby should be continous so charges will be taken

based on region and sex ==> groupby(["region","sex"])
and average of mean ===> ["charge"].mean()


```python
insurance_data.groupby(["region","sex"])["charges"].mean().round(2).sort_values(ascending=False)
```

## 2.Pivot Table

Pivot table is used to make the output presentable comparing two categorical(discrete) features and one continous data



```python
pd.pivot_table(data=insurance_data,values="charges",index="region",columns="sex").round(2)
```

## 3. Aggregate

Most frequently used aggregate function is Sum, min, max    

```python
insurance_data
```

## 3.1 Find out the average,min,max,sum insurance charges based on seperate regions and sex

here we are finding min,max,sum,average of insurances charge , by the column region and sex

**** Important ******
so to find out min,max,sum,average we are using function called AGGREGATE - agg()

```python
insurance_data.groupby(["region","sex"])["charges"].agg(func=["mean","min","max","sum"]).round(2)
```

## 6.1 Create a new column in the dataset where the column name should be eligiblity criteria and the condition is where age is > 20 - value should be eligble or not eligble

```python
insurance_data["age"]
```

```python
def get_age(x):
    if x > 20:
        return "eligible"
    else:
        return "not eligible"
```

```python
insurance_data["eligible criteria"] = insurance_data["age"].apply(get_age)
insurance_data
```

## 7. lambda function

Ananymous function or nameless function 

use single expression and return the value

```python
insurance_data["eligible criteria_2"] = insurance_data["age"].apply(lambda x:"eligible" if x > 20 else "not eligible")
insurance_data
```
