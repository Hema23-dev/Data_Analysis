## TASK 2 

MERGE FUNCTION IS USED TO COMBINE TWO DATAFRAMES BASED ON COMMON KEYS OR COLUMNS

TO COMBINE MORE THAN TWO DATAFRAME WE HAVE BUILT-IN FUNCTION functools.reduce() 

Properties
----------
1. we can combine two or more datasets
2. it is like SQL join mechanism where we can do it by 'how' parameter by giving values as below
    - how = left --> Keeps all the rows from the first dataset and matching rows from the second dataset
    - how = right --> Keeps all the rows from the second dataset and matching rows from the first dataset
    - how = inner --> Keeps only the rows where the key exists in both dataset. It is default 
    - how = outer --> keeps all the dataset from both the dataset and filling missing values as Nan

Merge Function 1 Example 

## Merge two or more dataset

-- chaining the pd.merge() function or using functools.reduce

```python
# chaing pd.merge()
merge_sales_2017_2018_2019 = pd.merge(pd.merge(sales_data_2017,sales_data_2018,how='inner'),sales_data_2019,how='inner')
merge_sales_2017_2018_2019
```

```python
merge_sales_2017_2018_2019.shape
```

```python
# functools.reduce

from functools import reduce

lst_sales_data = [ sales_data_2017 ,sales_data_2018 ,sales_data_2019 ]

merge_sales_2017_2018_2019_2 = reduce(lambda left,right : pd.merge(left,right,how="outer"),lst_sales_data)

merge_sales_2017_2018_2019_2 
```

## TASK 3 - LIST COMPREHENSION

List Comprehension - short way to create new list from an exisitng list or sequence

syntax: [expression for item in iterable]

```python
## squares of the number

#Normal code

lst_num = [1,2,3,4,5]
sq_lst_num = []

for i in lst_num:
    sq = i*i
    sq_lst_num.append(sq)
sq_lst_num
```

```python
# sq of num in list comprehension

sq_lst_num = [i*i for i in lst_num]

sq_lst_num
```

## list comprehension with if 

syntax : [expression for item in iterable if condition]

```python
## find the even number from list

## Normal code

lst_num = [1,2,3,4,5]

even_num = []

for i in lst_num:
    if i%2 == 0:
        even_num.append(i)
print(even_num)
```

```python
## find the even number from list

## list comprehension

even_num = [ i for i in lst_num if i%2 == 0 ]

print(even_num)

```

## list comprehension with if-else

syntax : [value_if_true if condition else value_if_false for item in iterable]


```python
## list the even and odd number from lst_num

result = [ "even"  if i % 2 == 0 else "odd" for i in lst_num ]

print(result)
```

## list comprehension with multiple conditions

syntax : [expression for item in iterable if condition1 if condition2]


```python
# print the even number greater than 5 from the list 

num = range(1,21)

res = [i for i in num if i%2 == 0 if i > 5]

print(res)
```

## list comprehension for Nested list to flatten 

syntax : [expression for outer_item in outer_iterable for inner_item in outer_item]


```python
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

flat_list = [ number for row in matrix for number in row]

print(flat_list)
```

## Given a list of sentences, use list comprehension to:

Extract all words.

Convert them to lowercase.

Keep only words with more than 4 characters.

Remove duplicate words.

Return the result in alphabetical order.

```python
## normal code

sentences = ["Python is powerful","List comprehension makes Python coding easier","Python is widely used"]
words=[]
for sentence in sentences:
    for word in sentence.split(): 
        word= word.lower()
        if len(word) > 4 and word not in words:
           words.append(word)

words.sort()
print(words)
        
```

```python
## list comprehension


out = sorted(set([word.lower() for sentence in sentences for word in sentence.split() if len(word) > 4 ]))  # using set to remove duplicates and sorted to sort the list


print(out)
```

## Given a list of employee records, use list comprehension to:

Select employees whose salary is at least 60000.

Convert each selected employee’s name to uppercase.

Return the result as a list of tuples containing the name and salary.

```python
employees = [
    ("Ravi", 55000),
    ("Anita", 72000),
    ("Kumar", 60000),
    ("Priya", 48000),
    ("Arun", 85000)
]

employees_filtered = []

for name,salary in employees:
    if salary >= 60000 :
        employees_filtered.append((name.upper(),salary))


print(employees_filtered)
    
```

```python
## list comprehension


employees_filtered = [(name.upper(),salary) for name,salary in employees if salary >= 60000]

print(employees_filtered)
```

## lambda map - applies a function to every item in the iterable such as list

syntax : map(function, iterable)

```python
## to retrun the square of the numbers from the list

num_lst = list(map(int,input("enter the number:").split()))

square_num = list(map(lambda x:x*x, num_lst))

print(square_num)
```

```python
## if we want get the list of input

lst = list(map(str,input("enter the name").split()))

print(lst)
```

```python
## convert the name in upper case

name_uppercase = list(map(lambda name:name.upper(),lst))
print(name_uppercase)
```

## filter() - selects only the items for which a condition returns True.

syntax - filter(function, iterable)

```python
## return the even number from the list

even_num = list(filter(lambda x:x%2==0 ,num_lst))  # here if the x%2 is true it returns the value to list , here the filter lambda will be useful

print(even_num)
```

## reduce() - repeatedly combines items and produces one final result. It is available in the functools module.

syntax - reduce(function, iterable)

```python
## to add the list of numbers

from functools import reduce

num_lst = list(map(int,input("enter the number:").split()))
  
lst_add = reduce(lambda x,y:x+y,num_lst)   # this add as (1+5->6+6->12+7->19+3=22)

print(lst_add)
```

## 1. we have a list of numbers. We want to do below operations using lambda

Keep only even numbers.

Square them.

Add the squared values.

```python
lst_no=[1,3,4,5,7,9,8]

even_no=list(filter(lambda x:x%2==0, lst_no))
print(even_no)
sqr_no=list(map(lambda x:x*x, even_no))
print(sqr_no)
add_sqr = reduce(lambda x,y:x+y,sqr_no)
print(add_sqr)
```

2. ## Write a Python program that: Calculates the average mark for every student using map() and lambda.

## Keeps only students whose average is at least 60 using filter() and lambda.

## Sorts the remaining students by average mark in descending order using sorted() and lambda.

## Displays the student name and average, rounded to two decimal places.

## Finds the highest average using reduce().


```python
students = [
    {"name": "Arun", "marks": [78, 85, 92]},
    {"name": "Bala", "marks": [45, 67, 52]},
    {"name": "Charan", "marks": [90, 88, 95]},
    {"name": "Divya", "marks": [60, 72, 68]},
    {"name": "Esha", "marks": [35, 40, 48]}
]
```

```python
## find average

stud_average = list(map(lambda student : {**student,"average": sum(student["marks"]) / len(student["marks"])},students)) ## student represents the student["name"] and student["marks"]

print(stud_average)
```

```python
## filter

pass_students = list(filter(lambda student: student["average"] >= 60 , stud_average))

print(pass_students)
```

```python
rank_students = sorted(passed_students,key=lambda student: student["average"],reverse=True)
print(rank_students)
```

```python
from functools import reduce

highest_average = reduce(lambda highest, student: max(highest, student["average"]),rank_students,0)

print(highest_average)
```
