```python
## 1.Create a list of numbers and print only even numbers using a for loop.

lst_num = [1,2,4,3,66,55,21,30,78,99,8,0,12,14,11]
even_num =[]
for i in lst_num:
    if i % 2 == 0:
       even_num.append(i)

print(f"the even number from the lst_num is ",even_num)
```

```python
## 2.Write a program that handles division by zero using try-except.


try:
   numerator = float(input("enter the numerator:"))
   denominator = float(input("enter the numerator:"))

   answer = numerator/denominator

   print(f"answer:{answer}")

except ZeroDivisionError:
   print("the zero division is not allowed")

except ValueError:
   print("Error: Please enter valid numeric value in denominator")
  
    
```

```python
## 3.Given a string, count how many vowels it has.

words = input("enter the string:")
vowels = "aeiou"
count = 0

for char in words.lower():
   if char in vowels:
      count = count + 1

print(count)
```

```python
## 4.Reverse a list without using built-in functions.

print(f"lst_num is {lst_num}")
rev_lst_num =[]

for i in range(len(lst_num)-1,-1,-1):
    rev_lst_num.append(lst_num[i])

print(f"rev_lst_num is {rev_lst_num}")
    
```

```python
## 8.Replace all missing values in a DataFrame with the column’s mean.

emp_data["Age"] = emp_data["Age"].convert_dtypes(convert_floating = False )
emp_data
```

```python
emp_data["Age"] = emp_data["Age"].fillna(int(round(emp_data["Age"].mean())))
```

```python
emp_data
```

```python
emp_data["Salary"]=emp_data["Salary"].fillna(emp_data["Salary"].mean())
emp_data
```
