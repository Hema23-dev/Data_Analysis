## Task

1. Find out the Top 5 selling items and individual revenue it made.

2. Total revenue of the Top 5 items sold


```python
order_details
```

```python
order_details
```

```python
top_5_items=order_details.groupby(by="item_name")["quantity"].sum().sort_values(ascending = False).reset_index(name="total_quantity").head(5)
top_5_items
```

```python
# Revenue for each order-detail row
order_details["revenue"] = (
    order_details["quantity"] * order_details["item_price"]
)

order_details
```

```python
# Find out the Top 5 selling items and individual revenue it made.

# .agg() --> means aggregate - summarize one or more dataframe such as total,average,min,max or count

top_5_items = (order_details.groupby("item_name", as_index=False).agg(total_quantity=("quantity", "sum"),revenue=("revenue", "sum"))
    .sort_values("total_quantity", ascending=False)
    .head(5)
)

top_5_items
```

```python
## Total revenue of the Top 5 items sold

top_5_total_revenue = top_5_items.agg(total_revenue = ("revenue","sum"))

top_5_total_revenue
```

```python
## fetching the chicken bowl's item quantity,revenue and item_price

## loc - select or update specific rows and columns in a pandas DataFrame , Syntax : dataframe.loc[rows, columns]

item_name = "Chicken Bowl"

item_chicken_bowl_details = order_details.loc[order_details["item_name"] == item_name,["item_name", "item_price", "quantity", "revenue"]]


item_chicken_bowl_details
```
